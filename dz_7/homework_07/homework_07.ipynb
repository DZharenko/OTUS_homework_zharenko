# I. Numpy

### Импортируйте NumPy

```python
import numpy as np
```

### Создайте одномерный массив размера 10, заполненный нулями и пятым элемент равным 1. Трансформируйте в двумерный массив.

```python
my_array = np.zeros(10)
my_array[4] = 1
my_array = my_array.reshape(2, 5)
my_array
```

### Создайте одномерный массив со значениями от 10 до 49 и разверните его (первый элемент становится последним). Найдите в нем все четные элементы.

```python
my_array = np.arange(10,50)
my_array = my_array[::-1]
even_elements = my_array[my_array % 2 == 0]
my_array
even_elements
```

### Создайте двумерный массив 3x3 со значениями от 0 до 8

```python
my_array = np.arange(0,9).reshape(3,3)
my_array
```

### Создайте массив 4x3x2 со случайными значениями. Найти его минимум и максимум.

```python
my_array = np.random.rand(4,3,2)
min_arr = my_array.min()
max_arr = my_array.max()
my_array, min_arr, max_arr
```

### Создайте два двумерных массива размерами 6x4 и 4x3 и произведите их матричное умножение. 

```python
my_array_1 = np.random.rand(6,4)
my_array_2 = np.random.rand(4,3)
result = my_array_1.dot(my_array_2)
my_array_1, my_array_2, result
```

Создайте случайный двумерный массив 7x7, найти у него среднее и стандартное оклонение. Нормализуйте этот массив.### 

```python
my_array = np.random.rand(7,7)
mean = my_array.mean()
std = my_array.std()
normalized_array = (my_array - mean)/std
normalized_array
```
